```python
# Re-run setup: libraries + load data
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

df = pd.read_csv("house_prices.csv")
print("Dataset loaded successfully")
print("Shape:", df.shape)
```

```python
# Part 2.1: Check Missing Values
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    "Missing Count": missing,
    "Missing Percentage (%)": missing_pct.round(2)
})
display(missing_report)

print("\nColumns with missing values:", missing[missing > 0].index.tolist() or "None")
```

### 2.2 Handling Missing Values
No missing values were found in any column, so no imputation (mean/median/mode) or row/column removal was required.

```python
# Part 2.3: Check Duplicate Records
# 'Unnamed: 0' is just an index, so drop it first to detect real duplicates
df = df.drop(columns=["Unnamed: 0"])

dup_count = df.duplicated().sum()
print("Duplicate records found:", dup_count)

before = df.shape[0]
df = df.drop_duplicates().reset_index(drop=True)
after = df.shape[0]

print("Rows before:", before)
print("Rows after :", after)
print("Duplicates removed:", before - after)
print("Remaining duplicates:", df.duplicated().sum())
```

```python
# Part 2.4: Check Data Types
print("Data types before correction:")
print(df.dtypes)

# Part 2.5: Rename Columns
df = df.rename(columns={
    "Area_in_Marla": "area_marla",
    "baths": "bathrooms",
    "price": "price_pkr"
})

# Convert text columns to category type
for col in ["property_type", "city", "purpose"]:
    df[col] = df[col].astype("category")

print("\nUpdated columns:", df.columns.tolist())
print("\nData types after correction:")
print(df.dtypes)
```

```python
print("Current shape:", df.shape)
print("Remaining duplicates:", df.duplicated().sum())
```

```python
# Part 2.6: Check Invalid Values
print("=== Summary Statistics ===")
display(df.describe())

print("\n=== Invalid Value Checks ===")
print("Price <= 0         :", (df["price_pkr"] <= 0).sum())
print("Bedrooms == 0      :", (df["bedrooms"] == 0).sum())
print("Bedrooms > 10      :", (df["bedrooms"] > 10).sum())
print("Bathrooms == 0     :", (df["bathrooms"] == 0).sum())
print("Bathrooms > 10     :", (df["bathrooms"] > 10).sum())
print("Area <= 0          :", (df["area_marla"] <= 0).sum())
print("Area > 100 Marla   :", (df["area_marla"] > 100).sum())

print("\n=== Categories ===")
print(df["purpose"].value_counts())
print(df["property_type"].value_counts())
print(df["city"].value_counts())
```

```python
# Part 2.6: Handle Invalid Values
print("Rows before cleaning:", df.shape[0])

# 1. Remove invalid area (<= 0)
df = df[df["area_marla"] > 0]

# 2. Remove zero-bedroom rows (likely data errors / not comparable)
df = df[df["bedrooms"] > 0]

# 3. Remove extreme area outliers (> 100 Marla)
df = df[df["area_marla"] <= 100]

# 4. Keep only 'For Sale' listings (rent and sale prices have different scales)
df_rent = df[df["purpose"] == "For Rent"].copy()
df = df[df["purpose"] == "For Sale"].copy()

# Reset index and clean unused category levels
df = df.reset_index(drop=True)
for col in ["property_type", "city", "purpose"]:
    df[col] = df[col].cat.remove_unused_categories()

print("Rows after cleaning :", df.shape[0])
print("Rent listings set aside:", df_rent.shape[0])
```

```python
# Part 2.7: Verify the Cleaned Dataset
print("=== Missing Values ===")
print(df.isnull().sum())

print("\n=== Duplicate Records ===")
print(df.duplicated().sum())

print("\n=== Data Types ===")
print(df.dtypes)

print("\n=== Dataset Shape ===")
print(df.shape)

print("\n=== Basic Statistics ===")
display(df.describe())
```

### Data Cleaning Summary
- **Missing values:** None found in any column, so no imputation was needed.
- **Index column:** The unnamed index column (`Unnamed: 0`) was dropped because it carried no information.
- **Duplicates:** 37,858 duplicate records (about 38% of the data) were found and removed. These were repeated scrapes of the same listing.
- **Renamed columns:** `price` to `price_pkr`, `baths` to `bathrooms`, `Area_in_Marla` to `area_marla`, for clarity and consistency.
- **Data types:** `property_type`, `city` and `purpose` were converted to the `category` type.
- **Invalid values:** 10 rows with area <= 0 were removed because zero area is impossible. 283 rows with 0 bedrooms were removed as they are not comparable. 41 rows with area above 100 Marla were removed as extreme outliers.
- **Rent vs Sale:** Rent listings (17,967 rows) were set aside because rent and sale prices are on completely different scales, and mixing them would distort the mean, correlation and charts. The analysis uses only "For Sale" listings (43,340 rows).

```python
# Part 3.1: Descriptive Statistics
num_cols = ["price_pkr", "bathrooms", "bedrooms", "area_marla"]

stats = pd.DataFrame({
    "Mean": df[num_cols].mean(),
    "Median": df[num_cols].median(),
    "Mode": df[num_cols].mode().iloc[0],
    "Min": df[num_cols].min(),
    "Max": df[num_cols].max(),
    "Range": df[num_cols].max() - df[num_cols].min(),
    "Std Dev": df[num_cols].std(),
    "Variance": df[num_cols].var(),
})
pd.options.display.float_format = "{:,.2f}".format
display(stats)
```

```python
# Part 3.2 - 3.5: Mean, Median, Mode, Standard Deviation
print("Mean price   : PKR {:,.0f}".format(df["price_pkr"].mean()))
print("Median price : PKR {:,.0f}".format(df["price_pkr"].median()))
print("Std deviation: PKR {:,.0f}".format(df["price_pkr"].std()))
print("Skewness     :", round(df["price_pkr"].skew(), 2))

print("\nMode of categorical columns:")
for col in ["property_type", "city", "location"]:
    print(f"  {col}: {df[col].mode()[0]}  ({df[col].value_counts().iloc[0]} listings)")

print("\nAverage price by city (PKR):")
print(df.groupby("city")["price_pkr"].agg(["mean", "median", "count"]).round(0).sort_values("mean", ascending=False))

print("\nAverage price by property type (PKR):")
print(df.groupby("property_type")["price_pkr"].agg(["mean", "median", "count"]).round(0).sort_values("mean", ascending=False))
```

### Statistical Analysis: Explanation

**Mean:** The average house price is about PKR 13.57 million. It gives a general idea of the market, but it is pulled upward by a small number of very expensive properties.

**Median:** The median price is PKR 10.9 million, lower than the mean. Because the data is right-skewed (skewness = 1.12), the median is a better measure of a "typical" property. A few luxury listings inflate the mean but do not affect the median.

**Mode:** The most frequent property type is House (29,041 listings), the most frequent city is Karachi (15,955 listings), and the most frequent location is DHA Defence (2,528 listings). Mode is useful for categorical variables where mean and median cannot be calculated.

**Standard Deviation:** The price standard deviation is about PKR 9.85 million, which is large compared to the mean. This shows that prices vary widely, from very cheap to very expensive properties.

**City comparison:** Islamabad and Lahore have the highest average prices (about 14.6 million), while Faisalabad has the lowest (about 10.7 million).

**Property type comparison:** Farm Houses have the highest average price and Rooms have the lowest, but both categories have very few listings (52 and 12), so their averages are less reliable. Houses and Flats make up most of the dataset.

```python
# Part 3.6: Correlation Analysis
corr = df[num_cols].corr()
display(corr.round(2))
```

### Correlation Analysis: Explanation

- **Area and price (0.61):** Area has the strongest relationship with price. Larger properties generally cost more, which is expected in real estate.
- **Bathrooms and price (0.58) and bedrooms and price (0.53):** Both show a moderate positive relationship. More rooms usually mean a bigger, more expensive house.
- **Bathrooms and bedrooms (0.83):** These two are very strongly related, because houses with more bedrooms almost always have more bathrooms. This is useful to know, as the two features carry similar information (multicollinearity), which matters for future machine learning tasks.
- **Area and bedrooms (0.38):** The weakest relationship. A large plot does not always mean more bedrooms, since area can also be used for gardens, parking or commercial space.
- No correlation is negative, and none is close to a perfect 1.0 (except with itself), so price depends on several factors together, not on a single feature. Location and city (categorical) also play a major role, which we will explore in the visualizations.

```python
# Setup for saving charts
import os
os.makedirs("images", exist_ok=True)
pd.options.display.float_format = None

def save_fig(name):
    plt.tight_layout()
    plt.savefig(f"images/{name}.png", dpi=150, bbox_inches="tight")
    plt.show()
```

```python
# Visualization 1: Histogram of Price
plt.figure(figsize=(10, 6))
sns.histplot(df["price_pkr"] / 1e6, bins=50, kde=True, color="steelblue")
plt.title("Distribution of House Prices (For Sale)", fontsize=14)
plt.xlabel("Price (Million PKR)")
plt.ylabel("Number of Listings")
save_fig("visualization_01_histogram_price")
```

### Visualization 1: Histogram - Interpretation
- **What it shows:** How house prices are distributed across all "For Sale" listings.
- **Pattern:** Most listings are priced between about 3 and 10 million PKR, with the highest peak around 4 to 8 million. The distribution has a long tail to the right, reaching about 45 million.
- **Meaning:** The distribution is right-skewed (skewness = 1.12). Affordable and mid-range properties are very common, while luxury properties are rare.
- **Why it matters:** This confirms why the mean (13.57 million) is higher than the median (10.9 million). The median is the better measure of a typical price.

```python
# Visualization 2: Bar Chart - Average Price by City
city_avg = df.groupby("city")["price_pkr"].mean().sort_values(ascending=False) / 1e6

plt.figure(figsize=(10, 6))
sns.barplot(x=city_avg.index, y=city_avg.values, palette="viridis")
plt.title("Average House Price by City", fontsize=14)
plt.xlabel("City")
plt.ylabel("Average Price (Million PKR)")
save_fig("visualization_02_bar_city_price")
```

### Visualization 2: Bar Chart - Interpretation
- **What it shows:** The average sale price of properties in each city.
- **Pattern:** Islamabad (about 14.6 million) and Lahore (about 14.6 million) have the highest averages, followed by Karachi (12.9 million) and Rawalpindi (12.1 million). Faisalabad is the cheapest (10.7 million).
- **Meaning:** The two largest and most developed markets, the capital and Lahore, are the most expensive, while Faisalabad is the most affordable.
- **Why it matters:** City is an important factor in pricing. A buyer on a limited budget gets more property in Faisalabad or Rawalpindi than in Islamabad or Lahore.

```python
# Visualization 3: Pie Chart - Property Type Share (fixed labels)
type_counts = df["property_type"].value_counts()
main = type_counts[type_counts > 1000]
other = type_counts[type_counts <= 1000].sum()
pie_data = pd.concat([main, pd.Series({"Other": other})])

plt.figure(figsize=(8, 8))
plt.pie(pie_data, labels=pie_data.index, autopct="%1.1f%%", startangle=90,
        colors=sns.color_palette("pastel"), pctdistance=0.75,
        wedgeprops={"edgecolor": "white"})
plt.title("Share of Listings by Property Type", fontsize=14)
save_fig("visualization_03_pie_property_type")
```

### Visualization 3: Pie Chart - Interpretation
- **What it shows:** The share of each property type among all "For Sale" listings.
- **Pattern:** Houses make up about two thirds of all listings (67%), followed by Flats (about 28%). All other types (Upper Portion, Lower Portion, Penthouse, Farm House, Room) together form only a small fraction.
- **Meaning:** The Pakistani sale market in this dataset is dominated by independent houses, with flats as the second most common choice.
- **Why it matters:** Any average price in this dataset is mostly describing houses and flats. Categories with very few listings, such as Room and Farm House, should not be used for strong conclusions.

```python
# Visualization 4: Scatter Plot - Area vs Price (sample for readability)
sample = df.sample(3000, random_state=42)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=sample, x="area_marla", y=sample["price_pkr"] / 1e6,
                hue="city", alpha=0.6)
plt.title("Property Area vs Price", fontsize=14)
plt.xlabel("Area (Marla)")
plt.ylabel("Price (Million PKR)")
save_fig("visualization_04_scatter_area_price")
```

### Visualization 4: Scatter Plot - Interpretation
- **What it shows:** The relationship between property area (Marla) and price, with each dot being one listing, coloured by city.
- **Pattern:** Most listings are below 20 Marla, and within this range price rises with area. There is a vertical line at exactly 20 Marla (1 Kanal), a very common standard plot size. Above 40 Marla there are very few listings and the relationship is scattered: some very large plots are not very expensive.
- **Meaning:** This matches the correlation of 0.61 between area and price. Area is the strongest numeric driver of price, but not the only one, because location and city also matter.
- **Why it matters:** Area alone cannot predict price. Properties of the same size differ widely in price depending on city and location, so a model would need those features too.

```python
# Visualization 5: Line Chart - Average Price by Number of Bedrooms
bed_avg = df.groupby("bedrooms")["price_pkr"].mean() / 1e6
plt.figure(figsize=(10, 6))
plt.plot(bed_avg.index, bed_avg.values, marker="o", linewidth=2, color="darkgreen")
plt.title("Average Price by Number of Bedrooms", fontsize=14)
plt.xlabel("Number of Bedrooms")
plt.ylabel("Average Price (Million PKR)")
plt.grid(True)
save_fig("visualization_05_line_bedrooms_price")
```

### Visualization 5: Line Chart - Interpretation
- **What it shows:** The average price for properties with 1 to 6 bedrooms. (The dataset has no date column, so a trend over bedrooms is used instead of a trend over time.)
- **Pattern:** The average price generally increases as the number of bedrooms increases.
- **Meaning:** More bedrooms usually mean a bigger and more expensive property, which is consistent with the correlation of 0.53 between bedrooms and price.
- **Why it matters:** Bedroom count is a useful pricing feature, but the trend can flatten or dip at the highest counts where there are fewer listings.

### Visualization 9: Pair Plot - Interpretation
